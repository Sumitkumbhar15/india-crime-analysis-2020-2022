```python
import pandas as pd
```

```python
# Load Data

data = pd.read_csv("../raw_data/statewise_crime_data.csv")
```

```python
# Remove the Serial Number

data = data.drop(columns = ["Sl. No."])
```

```python
# Rename the columns name

data.columns = ["State_UT" , "Crime_2020" , "Crime_2021" , "Crime_2022" , "Mid-Year_Projected_Population(2022)" , "IPC_2022" , "Chargesheeting_Rate(2022)"]
```

```python
# Remove the Total rows

data_clean = data[~data["State_UT"].str.contains("Total" , case=False)]

data_clean.tail()
```

```python
# Top 10 State or UT in Crimes (2022)

top_crimes = data_clean.sort_values(by="Crime_2022" , ascending=False).head(10)
top_crimes[["State_UT" , "Crime_2022"]]
```

```python
# Top 5 State or UT in Crime Rate (2022)

top_crime_rate = data_clean.sort_values(by="IPC_2022" , ascending=False).head(5)
top_crime_rate[["State_UT" , "IPC_2022"]]
```

```python
# state and UT's with low crime rate

low_crime_rate = data_clean.sort_values(by="IPC_2022").head(5)
low_crime_rate[["State_UT" , "IPC_2022"]]
```

```python
# Best Performing States on basis of Chargesheeting Rate

good_chargesheeting_rate = data_clean.sort_values(by="Chargesheeting_Rate(2022)").head(5)
good_chargesheeting_rate[["State_UT" , "Chargesheeting_Rate(2022)"]]
```

```python
# Worst Performing States on basis of Chargesheeting Rate

worst_chargesheeting_rate = data_clean.sort_values(by="Chargesheeting_Rate(2022)").tail(5)
worst_chargesheeting_rate[["State_UT" , "Chargesheeting_Rate(2022)"]]
```

```python
# Crime Increase / Decrease Percentage (2020 -> 2022)

data_clean = data_clean.copy()
data_clean.loc[: , "Crime_change(%)"] = (data_clean["Crime_2022"] - data_clean["Crime_2020"]) / data_clean["Crime_2022"] * 100
data_clean[["State_UT" , "Crime_2020" , "Crime_2022" , "Crime_change(%)"]]

data_clean["Crime_change(%)"] = data_clean["Crime_change(%)"].round(2)
data_clean["Crime_change(%)"] = data_clean["Crime_change(%)"].astype(int).astype(str) + "%"
```

```python
crime_percentage = data_clean[["State_UT" , "Crime_2020" , "Crime_2022" , "Crime_change(%)"]]
crime_percentage
```

```python
# Export Cleaned Data (CSV)

data_clean.to_csv("../processed_data/Cleaned_data.csv")
top_crimes.to_csv("../processed_data/top_crimes_states&UT.csv")
top_crime_rate.to_csv("../processed_data/top_crime_rate.csv")
low_crime_rate.to_csv("../processed_data/low_crime_rate.csv")
good_chargesheeting_rate.to_csv("../processed_data/good_chargesheeting_rate.csv")
worst_chargesheeting_rate.to_csv("../processed_data/worst_chargesheeting_rate.csv")
crime_percentage.to_csv("../processed_data/crime_percentage(2020-2022).csv")

# Export in Excel
with pd.ExcelWriter("../processed_data/Crime_Analysis_Report.xlsx") as writer:
    
    # Sheet 1: The full cleaned dataset
    data_clean.to_excel(writer, sheet_name='Full_Cleaned_Data')
    
    # Sheet 2: Top 10 State or UT in Crimes (2022)
    top_crimes.to_excel(writer, sheet_name='Top_10_Volume')
    
    # Sheet 3: Top 5 State or UT in Crime Rate (2022)
    top_crime_rate.to_excel(writer, sheet_name='Top_5_Crime_Rate')

    # Sheet 4: state and UT's with low crime rate
    low_crime_rate.to_excel(writer, sheet_name='Top_5_Crime_Rate')
    
    # Sheet 4: Chargesheeting Performance (Best & Worst)
    good_chargesheeting_rate.to_excel(writer, sheet_name='Police_Performance')
    worst_chargesheeting_rate.to_excel(writer, sheet_name='Police_Performance')
```
